```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/github_projects.csv")
```

```python
df.head()
```

```python
conn = sqlite3.connect("github_projects.db")
```

```python
df.to_sql("Repositories", conn, if_exists="replace", index=False)
```

```python
pd.read_sql_query("SELECT * FROM Repositories LIMIT 5;", conn)
```

```python
query = """
SELECT *
FROM Repositories
WHERE stars > 10000;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT *
FROM Repositories
WHERE name LIKE '%Machine%';
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT name, owner, stars, forks
FROM Repositories
WHERE stars > 10000
AND forks > 1000;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT name, owner, stars, forks
FROM Repositories
WHERE stars > 10000
OR forks > 5000;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT name, owner, language
FROM Repositories
WHERE NOT language = 'Python';
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT name, owner, stars
FROM Repositories
ORDER BY stars DESC
LIMIT 10;
"""

top10 = pd.read_sql_query(query, conn)
top10
```

```python
top10.shape
```

```python
query = """
SELECT COUNT(*) AS total_repositories
FROM Repositories;
"""

total_repositories = pd.read_sql_query(query, conn)
total_repositories
```

```python
query = """
SELECT AVG(stars) AS average_stars
FROM Repositories;
"""

average_stars = pd.read_sql_query(query, conn)
average_stars
```

```python
query = """
SELECT language, COUNT(*) AS repository_count
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 5
ORDER BY repository_count DESC;
"""

language_counts = pd.read_sql_query(query, conn)
language_counts
```

```python
plt.figure(figsize=(12, 6))

plt.barh(top10["name"], top10["stars"])

plt.xlabel("Stars")
plt.ylabel("Repository")
plt.title("Top 10 Most Popular Machine Learning Repositories")

plt.gca().invert_yaxis()

plt.show()
```

```python
df["created_date"] = pd.to_datetime(df["created_date"])
```

```python
df["created_year"] = df["created_date"].dt.year
```

```python
creation_trend = (
    df.groupby("created_year")
      .size()
      .reset_index(name="repository_count")
)
```

```python
creation_trend
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    creation_trend["created_year"],
    creation_trend["repository_count"],
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Number of Repositories")
plt.title("Machine Learning Repository Creation Trends")

plt.grid(True)

plt.show()
```

# Step 8 – Interpretation of Results

## Key Insights

### 1. Repository Popularity
The analysis shows that repository popularity is concentrated among a small number of projects. The Top 10 repositories have the highest number of stars, indicating that some Machine Learning projects receive significantly more attention from the GitHub community than others.

### 2. Repository Creation Trends
The repository creation trend shows that the number of repositories increased significantly between 2009 and 2018. The highest number of repositories in our dataset was recorded in 2018, with 18 repositories.

After 2018, the number of repositories decreased, with a temporary increase in 2020. By 2024, only one repository in our dataset was created.

### 3. Programming Languages
Grouping the repositories by programming language helps identify the languages most commonly used in the selected Machine Learning repositories. Languages with more than five repositories represent the most frequently occurring programming languages in our dataset.

### 4. Business Insight
The analysis can help a software company understand which Machine Learning repositories attract the most attention, which programming languages are commonly represented, and how repository creation has changed over time.

### 5. Data Limitation
The analysis is based on 100 repositories returned by the GitHub API search query. Therefore, the results represent the selected dataset and should not be considered a complete representation of all Machine Learning repositories on GitHub.
