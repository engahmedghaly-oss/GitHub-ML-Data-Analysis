```python
import requests
import pandas as pd
```

```python
url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"
```

```python
headers = {
    "Accept": "application/vnd.github+json",
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

print("Status Code:", response.status_code)
```

```python
data = response.json()

print(type(data))
print(data.keys())
```

```python
df = pd.DataFrame(data["items"])
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.dtypes
```

```python
df.info()
```

```python
df = df[
    [
        "name",
        "owner",
        "language",
        "stargazers_count",
        "forks_count",
        "watchers_count",
        "open_issues_count",
        "created_at",
        "updated_at",
        "license"
    ]
]
```

```python
df.columns
```

```python
df["owner"] = df["owner"].apply(
    lambda x: x["login"] if isinstance(x, dict) else None
)
```

```python
df["license"] = df["license"].apply(
    lambda x: x["name"] if isinstance(x, dict) else None
)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df["language"] = df["language"].fillna("Unknown")
```

```python
df["license"] = df["license"].fillna("No License")
```

```python
df.isnull().sum()
```

```python
df.duplicated(subset=["name", "owner"]).sum()
```

```python
df = df.drop_duplicates(subset=["name", "owner"])
```

```python
df["created_at"] = pd.to_datetime(df["created_at"])
df["updated_at"] = pd.to_datetime(df["updated_at"])
```

```python
df = df.rename(columns={
    "stargazers_count": "stars",
    "forks_count": "forks",
    "watchers_count": "watchers",
    "open_issues_count": "open_issues",
    "created_at": "created_date",
    "updated_at": "updated_date"
})
```

```python
df.to_csv("github_projects.csv", index=False)
```
